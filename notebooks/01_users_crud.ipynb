# 회원(users) CRUD

`shopping/users.py`의 함수로 회원 테이블을 생성·조회·수정·삭제합니다.

```python
import sys
sys.path.append("..")  # 프로젝트 루트의 shopping 패키지를 불러오기 위해

from pprint import pprint
from postgrest.exceptions import APIError
from shopping import users
```

## Create - 회원 추가

```python
new_user = users.create_user("kim@example.com", "김철수", "010-3333-3333")
pprint(new_user)
```

## Read - 전체 회원 / 이메일로 1명 조회

```python
pprint(users.get_users())
print("---")
pprint(users.get_user_by_email("kim@example.com"))
```

## Update - 휴대전화 변경

```python
updated = users.update_user_phone(new_user["id"], "010-9999-9999")
print("변경 후 phone:", updated["phone"])
```

## 제약조건 검증 - 같은 이메일로 다시 가입하면 UNIQUE 위반

```python
try:
    users.create_user("kim@example.com", "중복회원")
except APIError as e:
    print("가입 실패:", e.code, e.message)
```

## Delete - 회원 삭제

```python
print("삭제된 행 수:", users.delete_user(new_user["id"]))
print("삭제 후 조회:", users.get_user_by_email("kim@example.com"))
```
