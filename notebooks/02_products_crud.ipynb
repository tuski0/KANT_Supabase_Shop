# 상품(products) CRUD

`shopping/products.py`의 함수로 상품 테이블을 생성·조회·수정·삭제합니다.

```python
import sys
sys.path.append("..")  # 프로젝트 루트의 shopping 패키지를 불러오기 위해

from pprint import pprint
from postgrest.exceptions import APIError
from shopping import products
```

## Create - 상품 추가

```python
new_product = products.create_product("27인치 모니터", 250000, 5)
pprint(new_product)
```

## Read - 전체 상품 / 상품 1개 조회

```python
pprint(products.get_products())
print("---")
pprint(products.get_product(new_product["id"]))
```

## Update - 가격·재고 변경

```python
updated = products.update_product(new_product["id"], price=230000, stock=8)
print("변경 후:", updated["name"], updated["price"], "원 / 재고", updated["stock"])
```

## 제약조건 검증 - 가격은 0 이상 (CHECK)

```python
try:
    products.create_product("잘못된 상품", -1000)
except APIError as e:
    print("등록 실패:", e.code, e.message)
```

## Delete - 상품 삭제

```python
print("삭제된 행 수:", products.delete_product(new_product["id"]))
print("삭제 후 조회:", products.get_product(new_product["id"]))
```
